## LATIHAN 4

```python
import pandas as pd

df = pd.read_csv('data_kantin.csv')

print(df.isnull().sum())
df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong
```

Ini adalah bagian dari logika pembersihan data (Data Cleaning):  

- Kolom terjual (Numerik): Jika jumlah terjual kosong, sangat logis untuk berasumsi bahwa pada saat itu barang tersebut tidak laku sama sekali, jadi masuk akal jika kita menggantinya dengan angka 0.
- Kolom menu (Kategorikal): Jika nama menu kosong, data tersebut kehilangan identitasnya. Kita tidak bisa menebak menu apa itu, dan baris data tanpa nama barang menjadi tidak berguna untuk dianalisis lebih lanjut (tidak bisa dikelompokkan). Oleh karena itu, baris tersebut lebih baik dibuang (drop).

## LATIHAN 5

```python
print(df.duplicated().sum()) # jumlah baris duplikat
df = df.drop_duplicates()

df['harga'] = df['harga'].astype(int) # memastikan tipe data harga adalah integer
print(df.dtypes)
```

**JAWABAN:**

Memastikan tipe data benar sangat krusial karena komputer memproses data sesuai tipenya. Jika kolom harga secara tidak sengaja terbaca sebagai teks (string / object) alih-alih angka (integer), kamu tidak akan bisa menjumlahkan, mengalikan, atau mencari rata-ratanya. Fungsi astype(int) memastikan data tersebut diakui sebagai angka valid. Mengenai jumlah baris setelah drop_duplicates(), baris hanya akan berkurang jika ada dua baris (atau lebih) yang datanya sama persis dari ujung ke ujung.

## KERJA KELOMPOK - DATA LOADING

```python
import pandas as pd
df = pd.read_csv("data_kantin.csv")
print("Dataset berhasil dimuat.")
```

## KERJA KELOMPOK - DATA INSPECTION

```python
print("=== 5 DATA PERTAMA ===")
display(df.head())

print("\n=== INFORMASI DATASET ===")
df.info()

print("\n=== STATISTIK DESKRIPTIF ===")
display(df.describe())

print("\n=== UKURAN DATASET ===")
print("Jumlah baris dan kolom:", df.shape)
```

## KERJA KELOMPOK - MENGECEK KUALITAS DATA

```python
print("=== MISSING VALUE ===")
print(df.isnull().sum())

print("\n=== DATA DUPLIKAT ===")
print("Jumlah data duplikat:", df.duplicated().sum())

print("\n=== TIPE DATA ===")
print(df.dtypes)
```

## KERJA KELOMPOK - MENGUBAH TIPE TANGGAL DATA

```python
df["tanggal"] = pd.to_datetime(df["tanggal"], errors="coerce")

print(df.dtypes)
```

**JAWABAN :**

Kolom tanggal diubah menjadi tipe datetime agar dapat digunakan untuk analisis berdasarkan tanggal dan pengurutan data.

## KERJA KELOMPOK - MENGHAPUS DATA DUPLIKAT

```python
print("Jumlah data sebelum:", len(df))

df = df.drop_duplicates()

print("Jumlah data setelah:", len(df))
print("Duplikat yang dihapus:", 110 - len(df))
```

**JAWABAN :**

Data duplikat dihapus karena merupakan data yang sama dan dapat menyebabkan hasil perhitungan penjualan menjadi lebih besar dari kondisi sebenarnya.

## KERJA KELOMPOK - MENANGANI MISSING VALUE PADA MENU

```python
df.loc[
    df["menu"].isna() &
    (df["harga"] == 12000) &
    (df["kategori"] == "Makanan"),
    "menu"
] = "Nasi Goreng"

df.loc[
    df["menu"].isna() &
    (df["harga"] == 3000) &
    (df["kategori"] == "Minuman"),
    "menu"
] = "Teh Hangat"

df.loc[
    df["menu"].isna() &
    (df["harga"] == 4000) &
    (df["kategori"] == "Minuman"),
    "menu"
] = "Es Teh"

print("Missing value pada menu:", df["menu"].isna().sum())
```

**JAWABAN :**

Missing value pada menu tidak langsung dihapus karena nama menu masih dapat ditentukan berdasarkan kombinasi harga dan kategori.

## KERJA KELOMPOK - MENANGANI MISSING VALUE TERJUAL

```python
median_menu = df.groupby("menu")["terjual"].median()

print("Median terjual setiap menu:")
display(median_menu)

df["terjual"] = df["terjual"].fillna(
    df["menu"].map(median_menu)
)

print("\nJumlah missing value terjual setelah cleaning:")
print(df["terjual"].isna().sum())
```

**JAWABAN :**

Median berdasarkan menu digunakan karena jumlah penjualan tiap menu berbeda. Cara ini lebih sesuai daripada menggunakan satu median untuk seluruh dataset.

## KERJA KELOMPOK - MEMASTIKAN TIPE DATA SUDAH SESUAI

```python
df["terjual"] = df["terjual"].round().astype(int)

print("=== TIPE DATA SETELAH CLEANING ===")
df.info()

print("\n=== MISSING VALUE SETELAH CLEANING ===")
print(df.isnull().sum())

print("\n=== DUPLIKAT SETELAH CLEANING ===")
print("Jumlah duplikat:", df.duplicated().sum())
```

##KERJA KELOMPOK - FILTERING

```python
df_filter = df[df["terjual"] >= 20]

print("Data penjualan dengan jumlah terjual >= 20:")
display(df_filter)
```

##KERJA KELOMPOK - SORTING

```python
df_sort = df.sort_values(
    by="terjual",
    ascending=False
)

print("10 penjualan tertinggi:")
display(
    df_sort[[
        "tanggal",
        "menu",
        "kategori",
        "harga",
        "terjual"
    ]].head(10)
)
```

##KERJA KELOMPOK - MEMBUAT KOLOM TURUNAN

```python
df["pendapatan"] = df["harga"] * df["terjual"]

print("Data dengan kolom pendapatan:")
display(df.head())
```

##KERJA KELOMPOK - GROUPY / AGREGASI

```python
analisis_menu = df.groupby("menu").agg(
    total_terjual=("terjual", "sum"),
    total_pendapatan=("pendapatan", "sum"),
    rata_rata_terjual=("terjual", "mean")
).sort_values(
    by="total_pendapatan",
    ascending=False
)

print("Analisis penjualan berdasarkan menu:")
display(analisis_menu)
```
